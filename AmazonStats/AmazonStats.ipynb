### Importing and Cleaning Data

Imporing pandas and accessing the csv file, with example head.

```python
import pandas as pd
amazon = pd.read_csv('amazon.csv')
amazon.head()
```

Getting DataFrame information (column names, types, etc)

```python
amazon.info()
```

We want to examine some numerical statistics about products based on category, rating, discount etc. We can remove columns relating to the users, reviews and links. We then also want to examine the missing rows in the rating_count column to see what has occurred.

```python
drop_columns = [amazon.columns[i] for i in range(8,16)]
amazon = amazon.drop(labels=drop_columns, axis=1)
```

```python
amazon[amazon.isnull().any(axis=1)]
```

These appear to simply be missing data, so we will also omit these rows.

```python
amazon=amazon[(amazon.isnull().any(axis=1))==False]
```

One of the rating values also appears to be a non-number, so will need to be removed.

```python
amazon.sort_values(by='rating',ascending=False).head(1)
```

For simpler analysis, we would also like to change the discount_percentage values to be float64 type with a decimal representation. We also change the price and rating_count values to be int64, and rating to float64.

```python
#Remove the above row
amazon=amazon.drop([1279])

#Function to remove non-number characters in a string
def getNumber(inp):
    out = ""
    inp=str(inp)
    for char in inp:
        if char.isnumeric():
            out += char
    return int(out)

#Remove non-number characters and assign proper column data types
amazon = amazon[['product_id','product_name','category','discount_percentage','discounted_price','actual_price','rating','rating_count']] 
for i in range(3,7):
    amazon[amazon.columns[i]] = amazon[amazon.columns[i]].apply(getNumber).astype('int64')
amazon['rating_count']=amazon['rating_count'].apply(getNumber).astype('int64')
amazon['rating']=amazon['rating']/10
amazon['discount_percentage']=amazon['discount_percentage']/100
```

We assign the cleaned amazon DataFrame to the new clean_amazon DataFrame.

```python
clean_amazon = amazon
clean_amazon.head()
```

### Analysis of Data

To examine some trends in the data, we will mainly use the rating values as a metric. We therefore may want to sort some of the input variables into discrete categories for easier comparison (such as bands of discounts separated by 0.10 or ratings counts separated by 5000). We add these as new columns in the DataFrame.

```python
def discountBand(discount):
    for i in range(10,101,10):
        if discount*100 < i:
            return str(i-10)+"-"+str(i) + "%"

clean_amazon['discount_band']=clean_amazon['discount_percentage'].apply(discountBand)

def ratingBand(rating_count,step):
    i = step
    while rating_count > i:
        i += step
    return i

clean_amazon['rating_band']=clean_amazon['rating_count'].apply(ratingBand,step=5000)
```

We can start by finding the mean rating of data grouped by these two data bands.

```python
discount_band_means=clean_amazon.groupby('discount_band').agg({'product_id':'nunique','rating':'mean'}).reset_index()
discount_band_means.columns=['discount_band','item_count','mean_rating']
discount_band_means
```

Perhaps unsurprisingly, there does not appear to be a strong correlation between the discount percentage and the ratings. This could be because the discount changes over time, and so ratings could come from periods where the items are not discounted or at a different band. The band with the most items is the 50-60% band, which is a little surprising as we might expect the majority of items to not be discounted or only a small discount. This could indicate that the data sample is not representative, maybe coming from a sample of specifically discounted items.

Next we can look at the data grouped by the rating count bands:

```python
rating_band_means=clean_amazon.groupby('rating_band').agg({'product_id':'nunique','rating':'mean'}).reset_index()
rating_band_means.columns=['rating_band','item_count','mean_rating']
rating_band_means
```

Firstly, we can note that the majority of the data falls in the first few groups. This indicates we may see better differences and trends in the data if we use smaller bands in the first 20000-30000 rating count items and either group or discard the outliers. For brevity, we will discard the outliers. 

```python
rating_band_clean = clean_amazon
rating_band_clean=rating_band_clean[rating_band_clean['rating_count']<25000]
rating_band_clean['rating_band']=rating_band_clean['rating_count'].apply(ratingBand,step=2500)

rating_band_clean_means=rating_band_clean.groupby('rating_band').agg({'product_id':'nunique','rating':'mean'}).reset_index()
rating_band_clean_means.columns=['rating_band','item_count','mean_rating']
rating_band_clean_means
```

This is perhaps a more representative view of the items in the sample, but again unfortunately doesn't point to many larger trends. Most ratings average around 4, with perhaps the items with the most reviews having slightly lower average ratings and those with fewer having higher. This could also be seen in the full data with the outliers. This could be due to people being more likely to leave a review if a product is faulty or misleading, so products with many reviews have lower averages. On the other end, products with few reviews may skew more positive as only people who were particularly pleased with the product bothered to leave a review (maybe an example of a Response Bias where responses tend to skew to extremes).
